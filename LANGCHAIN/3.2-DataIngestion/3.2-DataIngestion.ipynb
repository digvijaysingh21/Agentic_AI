```python
import wikipedia
from langchain_community.document_loaders import WikipediaLoader

# # Identify your application to Wikimedia
# wikipedia.set_user_agent(
#     "LangChainLearningProject/1.0 (educational use)"
# )

# Load Wikipedia content through LangChain
loader = WikipediaLoader(
    query="Generative AI",
    lang="en",
    load_max_docs=2,
    doc_content_chars_max=5000,
)

docs = loader.load()

print("Number of documents:", len(docs))

for doc in docs:
    print("\nTitle:", doc.metadata.get("title"))
    print("Summary:", doc.metadata.get("summary"))
    print("\nContent:\n", doc.page_content[:1000])
```
